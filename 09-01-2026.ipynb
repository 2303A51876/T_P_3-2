```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def addAtBeginning(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def addAtEnd(self, data):
        new_node = Node(data)
        if self.head == None:
            self.head = new_node
        else:
            current = self.head
            while current.next:
                current = current.next
            current.next = new_node

    def addAtPostion(self, data, position):
        new_node = Node(data)
        if position <= 0:
          self.addAtBeginning(data)
        elif position >= self.length():
          self.addAtEnd(data)
        else:
          current = self.head
          for _ in range(position - 1):
            current = current.next
          new_node.next = current.next
          current.next = new_node

    def length(self):
        count = 0
        current = self.head
        while current:
            count += 1
            current = current.next
        return count

    def display(self):
        current = self.head
        while current:
            print(current.data, end=" -> ")
            current = current.next
        print("None")

sll = SinglyLinkedList()
sll.addAtBeginning(30)
sll.addAtBeginning(20)
sll.addAtBeginning(10)
sll.addAtEnd(40)
sll.addAtPostion(25, 2)
print(sll.length())
sll.display()
```
